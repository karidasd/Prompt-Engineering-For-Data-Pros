## 1. Zero-Shot vs Few-Shot Learning
Let's see how adding examples changes the output structure for parsing unstructured logs.

```python
def extract_log_data(log_message, use_few_shot=False):
    system_prompt = "Extract the date, time, error level, and message from the log into JSON."
    
    if use_few_shot:
        system_prompt += """\n
        Example 1:
        Input: '2023-10-01 12:00:00 ERROR DB Crash'
        Output: {"date": "2023-10-01", "time": "12:00:00", "level": "ERROR", "msg": "DB Crash"}
        """
        
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        temperature=0.0,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": log_message}
        ]
    )
    return response.choices[0].message.content

test_log = "2024-05-12 08:33:21 WARN Memory threshold exceeded in worker_node_3"
print("ZERO SHOT OUTPUT:\n", extract_log_data(test_log, use_few_shot=False))
print("\nFEW SHOT OUTPUT:\n", extract_log_data(test_log, use_few_shot=True))
```

## 2. Chain of Thought Reasoning
Forcing the model to explain its steps before giving the final SQL query.

```python
task = "Write a SQL query to find the 2nd highest salary from the Employee table."

response = client.chat.completions.create(
    model="gpt-4",
    messages=[
        {"role": "system", "content": "You are a SQL Expert."},
        {"role": "user", "content": f"{task}\n\nLet's think step by step before writing the query."}
    ]
)
print(response.choices[0].message.content)
```
